```python
df["date"]=pd.to_datetime(df["date"],utc=True)
```

```python
df['date'].dtype
```

```python
df=df.sort_values("date")
```

```python
df
```

```python
#OHLC
```

```python
plt.figure(figsize=(12,8))
plt.plot(df.date,df["open"],label="open")
plt.plot(df.date,df["high"],label="high")
plt.plot(df.date,df["low"],label="low")
plt.plot(df.date,df["close"],label="close")

plt.title("Shopify stock Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.plot(df.date,df["volume"],label="volume")

plt.title("Shopify stock volume")
plt.xlabel("Date")
plt.ylabel("volume")
plt.legend()
plt.xticks(rotation=45)
```

```python
df["MA20"]=df["close"].rolling(window=20).mean()
df["MA50"]=df["close"].rolling(window=50).mean()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df.index,df["close"],label="Daily Close")
plt.title("Shopify Stock Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(14,7))
plt.plot(df.date,df["MA20"],label="MA20")
plt.plot(df.date,df["MA50"],label="MA50")
plt.plot(df.date,df["close"],label="close")

plt.title("Shopify Stock Price")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.legend()
plt.show()
```

```python
df["Daily_Return"]=df["close"].pct_change()
df["Daily_Return"].dropna()
```

```python
plt.figure(figsize=(14,7))
sns.histplot(
    data=df,
    x="Daily_Return",
    kde=True,
    bins=50
)
```

```python
df["Rolling_Volatility_20"]=(df["Daily_Return"].rolling(window=20).std())
df["Rolling_Volatility_20"].dropna()
```

```python
plt.figure(figsize=(12,8))
plt.plot(df.date,df["Rolling_Volatility_20"],label="Rolling_Volatility_20")
plt.title("Shopify Stock Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
```
